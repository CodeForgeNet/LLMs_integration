```python
from huggingface_hub import InferenceClient
import config
```

- huggingface_hub is a Python library by Hugging Face that lets you interact with their Model Hub, where thousands of AI/ML models (like GPT, BERT, Stable Diffusion, etc.) are hosted.
- InferenceClient is a class provided by that library.
- It’s basically a ready-made client that allows you to send text, image, or audio inputs to a Hugging Face model and get inference results (predictions) back.

```python
hf_token = config.HF_TOKENS
```

```python
# Initialize client for Mistral-7B-Instruct
client = InferenceClient(
    "mistralai/Mistral-7B-Instruct-v0.2",
    token=hf_token
)
```

## Generating Text

🔑 Key points
- Use chat_completion for instruct/chat-tuned models like Mistral-7B-Instruct.
- Use text_generation only for plain decoder models (like GPT-2, Falcon-7B base, etc.).

```python
# Function for generation
def generate_text(prompt):
    response = client.chat_completion(
        messages=[{"role": "user", "content": prompt}],
        max_tokens=50,
        temperature=0.7
    )
    # Response is a ChatCompletion object
    return response.choices[0].message["content"].strip()
```

```python
prompt = "A long time ago"
```

```python
generated_text = generate_text(prompt)
print(prompt, generated_text)
```

### Customizing the Output

⚡ Key fixes:
- Use chat.completions.create instead of text_generation.
- Wrap your input in messages=[{"role": "user", "content": prompt}] because Hugging Face runs these models in chat format.
- Access the text with response.choices[0].message["content"].

* text_generation → “Continue this text like autocomplete.”
* chat.completions.create → “Answer this user as if we’re chatting, with system/user roles.”

```python
def generate_text(prompt, max_tokens):
    response = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt}],
        max_tokens=max_tokens,
        temperature=0.7,
    )
    return response.choices[0].message["content"].strip()
```

```python
prompt = "A long time ago"
```

```python
generated_text = generate_text(prompt, max_tokens=5)
print(prompt, generated_text)
```

```python
generated_text = generate_text(prompt,20)
print(prompt, generated_text)
```

```python
# Temperature varies 0 to 1. 0 is less random, 1 is most random
def generate_text(prompt, max_tokens,temperature):
    response = client.chat.completions.create(
        messages=[{"role": "user", "content": prompt}],
        max_tokens=max_tokens,
        temperature=temperature,
    )
    return response.choices[0].message["content"].strip()
```

```python
prompt = "A long time ago"
```

```python
generated_text = generate_text(prompt, 30, 0)
print(prompt, generated_text)
```

```python
generated_text = generate_text(prompt, 30, 1)
print(prompt, generated_text)
```
